```python
# Importem les llibreries necessàries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Carreguem el dataset
df = pd.read_csv("steam.csv")

# Mostrem les primeres files
df.head()
```

```python
# Dimensions del dataset
print("Nombre de files i columnes:")
print(df.shape)

# Nom de les columnes
print("\nColumnes del dataset:")
print(df.columns)

# Tipus de dades
print("\nTipus de dades:")
print(df.dtypes)
```

```python
# Comprovem valors nuls
print("Valors nuls per columna:")
print(df.isnull().sum())
```

```python
# Estadístiques descriptives de les variables numèriques
df.describe()
```

```python
plt.figure(figsize=(10,5))
plt.hist(df['price'], bins=30)
plt.title('Distribució dels preus dels videojocs')
plt.xlabel('Preu (€)')
plt.ylabel('Nombre de videojocs')
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.hist(df['positive_ratings'], bins=30)
plt.title('Distribució de valoracions positives')
plt.xlabel('Valoracions positives')
plt.ylabel('Nombre de videojocs')
plt.show()
```

```python
plt.figure(figsize=(8,4))
plt.boxplot(df['price'])
plt.title('Diagrama de caixa del preu')
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.scatter(df['price'], df['positive_ratings'], alpha=0.5)
plt.title('Preu vs Valoracions positives')
plt.xlabel('Preu (€)')
plt.ylabel('Valoracions positives')
plt.show()
```

```python
top10 = df.sort_values(by='positive_ratings', ascending=False).head(10)

plt.figure(figsize=(12,6))
plt.bar(top10['name'], top10['positive_ratings'])
plt.title('Top 10 videojocs amb més valoracions positives')
plt.xlabel('Videojoc')
plt.ylabel('Valoracions positives')
plt.xticks(rotation=45, ha='right')
plt.show()
```
